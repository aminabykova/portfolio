# **AdvancedVocab API**
## Приложение-помощник для сохранения и тренировки новых английских слов
Прототип API-сервиса, предназначенного для продвинутого изучения английского языка. Сервис позволяет пользователям сохранять незнакомые слова, дополняя их определениями, синонимами, а также личными ассоциациями через цветовую маркировку. Основная цель — систематизировать процесс запоминания редкой лексики, сделать его более интерактивным и эффективным за счёт контекстуализации и мнемонических техник.

### Введение
Когда я достигла знания английского языка на продвинутом уровне, изучение и запоминание новых слов стало сложнее и медленнее, так как слова уровня C1-C2 в целом редко встречаются, а повторно — тем более, потому что в жизни нет полноценной ежедневной практики языка. А совершенствовать свой вокабуляр мне всё равно очень хочется.

Мне важна скорость записи новых слов, так как обычно я встречаю их, читаю что-то на телефоне, где есть возможность нажать на слово и быстро посмотреть его значение во встроенном словаре, не переходя в другие приложения. Но раньше сохранить эти знания я могла только сделав скриншот, который потом навсегда терялся в фотоплёнке, поэтому какое-то время назад я создала для себя тг-канал, куда отправляла эти скриншоты, не засоряя галерею, но проблема низкой читаемости этих картинок всё ещё осталась. Одной из возможных реализаций (северной звездой) более эффективного инструмента я вижу телеграм-бот, в который я буду отправлять только слово, а всё остальное (в т.ч. описанное ниже) будет выполнять он, но эта часть останется за рамками этого проекта.

### Идея сервиса
Моей гипотезой является то, что если у меня появится база знаний с этими словами, их значениями, синонимами и примерами использования, то шансов их запомнить будет больше, потому что они будут систематизированы, а так же у меня будет возможность кастомизировать хранимую информацию, а именно — присваивать словам один из 8 цветов (белый, серый, чёрный, красный, жёлтый, зелёный, синий, розовый), тем самым добавить в мой словарь немного синестезии.

Одним из важных для меня аспектов для запоминания слов является составление предложений с ними, поэтому это будет одним из способов закрепления каждого слова (составлять предложение не обязательно в моменте поиска значения, наоборот, лучше делать это в свободное время без спешки, когда когнитивных ресурсов больше). Дополнительной функциональностью приложения я хотела бы видеть LLM, которая составляла бы для меня одно предложение из нескольких (3х) слов в БД, так как связанной историей запоминать слова проще.

Ассоциации для цветов, присваеваемых словам:
* white: pure, young
* grey: undefined, neutral
* black: negative, dark
* red: negative, bad
* yellow: good, bright
* green: nice, everyday, earthly
* blue: structured, serious, neutral
* pink: magical, mysterious, feminine, cute

### Job Stories


1. **Когда я** встречаю незнакомое слово и ищу его значение, **я хочу** быстро записать информацию о нём в свою базу знаний, **чтобы** иметь возможность обратиться к нему позже и повторить.

2. **Когда я** выделяю время на повторение новых слов, **я хочу** составлять с ними предложения на основе определения, синонимов и присвоенного цвета, **чтобы** их смысл лучше запоминался.

3. **Когда я** хочу проверить свои знания в новом контексте, **я хочу** получать новые сгенерированные предложения, включающие несколько случайных слов из моего «словаря», **чтобы** тренировать их использование на новых примерах в связной речи.

4. **Когда я** подвожу итоги года, **я хочу** иметь возможность посмотреть статистику изученных/записанных в «словарь» слов, **чтобы** узнать, какой месяц был самым продуктивным и какие незнакомые слова (каких частей речи/цветов) встречались мне чаще всего.

5. **Когда я** сталкиваюсь с трудностью запоминания редких слов, **я хочу** фильтровать их по цветовым меткам, **чтобы** создать дополнительные мнемонические связи и использовать эффект двойной кодировки: текстовая информация (значения) связывается с визуальной (цвет), повышая вероятность долгосрочного удержания в памяти.

### Концептуальная модель
#### **Объекты и их атрибуты:**


*   Слово *(Word):*

    *   определение *(definition)*
    *   часть речи *(part_of_speech)*
    *   синонимы *(synonyms)*
    *   цветовая метка *(color)*
    *   предложение, составленное пользователем *(my_sentence)*
    *   дата добавления слова *(date_added)*

*   Предложения от ИИ *(AI_story)* (в прототипе реализована как заглушка):

    *   использованные слова *(used_words)*
    *   дата генерации *(gen_date)*
* Отчёт *(Report):*
    *   дата генерации *(report_date)*

#### **Действия:**


*   Слова:
    * **найти** (найти определение и синонимы слова без сохранения в БД или найти в своём словаре)
    * **записать** (сохранить новое слово и информацию о нём в БД.)
    * **изменить** (добавить/изменить цветовую метку или пользовательское предложение для существующего слова)
*   Предложения от ИИ-агента:
    * **сгенерировать** (запросить "умную" историю, использующую случайные слова из словаря и сохранить её в БД)
* Отчёт:
    * **сгенерировать** (сгенерировать годовой отчет по активности)

### Инициализация Flask приложения

```python
app = Flask(__name__)

@app.before_request
def before_request_func():
    """
    Эта функция выполняется перед КАЖДЫМ запросом к API.
    Она гарантирует, что соединение с базой данных установлено.
    """
    connect_to_db()

@app.teardown_appcontext
def teardown_db(exception):
    """
    Эта функция выполняется после КАЖДОГО запроса, даже если он завершился с ошибкой.
    Она закрывает соединение с базой данных, чтобы не оставлять "висящих" подключений.
    """
    disconnect()
```

### Реализация API эндпоинтов

```python
@app.route("/")
def home():
    """Домашняя страница с описанием API."""
    return """
    <h1>Welcome to AdvancedVocab API!</h1>
    <p>This is a tool to help you learn and practice advanced English vocabulary.</p>
    <h2>Available Endpoints:</h2>
    <ul>
        <li><b>GET /search?word=&lt;word&gt;</b>: Search for a word's definition and synonyms without saving.</li>
        <li><b>POST /words/&lt;word&gt;</b>: Add a new word to your dictionary. Requires JSON body: {"part_of_speech": "noun"}.</li>
        <li><b>GET /words</b>: Get a list of all words in your dictionary.</li>
        <li><b>GET /words/&lt;word&gt;</b>: Get all details for a specific word.</li>
        <li><b>PUT /words/&lt;word&gt;</b>: Update a word's color or your personal sentence. Requires JSON body: {"color": "blue", "my_sentence": "..."}.</li>
        <li><b>GET /filter/color/&lt;color&gt;</b>: Get all words with a specific color.</li>
        <li><b>GET /filter/color/&lt;color&gt;/pos/&lt;part_of_speech&gt;</b>: Filter words by color and part of speech.</li>
        <li><b>GET /generate-story</b>: [STUB] Get a sample story using 3 random words from your dictionary.</li>
        <li><b>GET /report/yearly</b>: [JINJA REPORT] Get a yearly progress report.</li>
    </ul>
    """

# Функция 1: Использование внешнего API
@app.route("/search", methods=['GET'])
def search_word_external():
    """Ищет слово, используя только внешние API, не сохраняя в БД."""
    word_to_find = request.args.get('word')
    if not word_to_find:
        return jsonify({"error": "Query parameter 'word' is required."}), 400

    definition = get_definition(word_to_find)

    return jsonify({
        "word": word_to_find,
        "definition": definition,
    })

# Функция 2: Комплексная функция (Внешний API + Запись в БД)
@app.route("/words/<string:word_text>", methods=['POST'])
def add_new_word(word_text):
    """Добавляет новое слово в базу данных."""
    word_text = word_text.lower()

    # Проверка, есть ли слово в базе
    if Word.objects(word=word_text).first():
        return jsonify({"message": f"Word '{word_text}' already exists in your dictionary."}), 409

    data = request.get_json()
    if not data or 'part_of_speech' not in data:
        return jsonify({"error": "JSON body with 'part_of_speech' is required."}), 400

    definition = get_definition(word_text)

    new_word = Word(
        word=word_text,
        part_of_speech=data['part_of_speech'],
        definition=definition,
        synonyms=[],
        date_added=datetime.utcnow()
    )
    new_word.save()

    return jsonify({
        "message": f"Word '{word_text}' successfully added to your dictionary.",
        "data": new_word.to_dict()
    }), 201

# Другие полезные методы для работы со словами
@app.route("/words", methods=['GET'])
def get_all_words():
    """Возвращает список всех сохраненных слов."""
    words = Word.objects.scalar('word') # .scalar('word') возвращает только значения поля word
    return jsonify({"words": list(words)})

@app.route("/words/<string:word_text>", methods=['GET'])
def get_word_details(word_text):
    """Возвращает полную информацию о слове из БД."""
    word_obj = Word.objects(word=word_text.lower()).first()
    if not word_obj:
        return jsonify({"error": "Word not found in your dictionary."}), 404
    return jsonify(word_obj.to_dict())

@app.route("/words/<string:word_text>", methods=['PUT'])
def update_word(word_text):
    """Обновляет цвет, предложение или список синонимов для существующего слова."""
    word_obj = Word.objects(word=word_text.lower()).first()
    if not word_obj:
        return jsonify({"error": "Word not found in your dictionary."}), 404

    data = request.get_json()
    if not data:
        return jsonify({"error": "Request body cannot be empty."}), 400

    updated = False
    if 'color' in data:
        word_obj.color = data['color']
        updated = True
    if 'my_sentence' in data:
        word_obj.my_sentence = data['my_sentence']
        updated = True
    if 'synonyms' in data:
        # Убеждаемся, что на вход пришел именно список
        if isinstance(data['synonyms'], list):
            word_obj.synonyms = data['synonyms']
            updated = True
        else:
            return jsonify({"error": "'synonyms' must be a list of strings."}), 400

    if not updated:
        return jsonify({"message": "No valid fields to update were provided."}), 400

    word_obj.save()
    return jsonify({
        "message": f"Word '{word_text}' has been updated.",
        "data": word_obj.to_dict()
    })

@app.route("/filter/color/<string:color>", methods=['GET'])
def filter_by_color(color):
    """Фильтрует слова по цвету."""
    words = Word.objects(color=color.lower())
    result = [word.to_dict() for word in words]
    return jsonify(result)

@app.route("/filter/color/<string:color>/pos/<string:part_of_speech>", methods=['GET'])
def filter_by_color_and_pos(color, part_of_speech):
    """Фильтрует слова по цвету и части речи."""
    words = Word.objects(color=color.lower(), part_of_speech=part_of_speech.lower())
    result = [word.to_dict() for word in words]
    return jsonify(result)

# Функция 3: Заглушка (симуляция ИИ)
@app.route("/generate-story", methods=['GET'])
def generate_story_stub():
    """
    Заглушка для генерации истории.
    Выбирает 3 случайных слова из БД и возвращает одну из заранее заготовленных историй.
    Симулирует работу сложной LLM, которая могла бы связать эти слова в рассказ.
    """
    all_words = list(Word.objects.scalar('word'))
    if len(all_words) < 3:
        return jsonify({"error": "Not enough words in dictionary to generate a story. Add at least 3 words."}), 40_0

    used_words = random.sample(all_words, 3)

    canned_stories = [
        "The nascent tech startup struggled to uphold its reputation after a competitor spread derogatory rumors about its founder.",
        "She had the gumption to gainsay her boss’s decision, despite the caveat that it might cost her the promotion.",
        "The murky waters near the docks were a blight on the otherwise laudable efforts to clean up the river.",
        "He would sporadically dabble in painting, though his crudely executed works were far from the paragon of art.",
        "The team’s benevolent leader refused to gloat over their victory, even when rivals went astray with their strategies."
    ]

    story = random.choice(canned_stories)

    return jsonify({
        "message": "[STUB] This is a simulated story. An advanced AI model would create a unique story from your words.",
        "words_chosen": used_words,
        "generated_story": story
    })

# Функция 4: Отчёт JINJA
@app.route("/report/yearly", methods=['GET'])
def yearly_report():
    """
    Генерирует и возвращает HTML-отчет о прогрессе за год с использованием Jinja2.
    """
    # HTML-шаблон для отчёта
    JINJA_TEMPLATE = """
    <!DOCTYPE html>
    <html lang="en">
    <head>
        <meta charset="UTF-8">
        <title>Yearly Vocabulary Report</title>
        <link rel="stylesheet" href="https://stackpath.bootstrapcdn.com/bootstrap/4.4.1/css/bootstrap.min.css" integrity="sha384-Vkoo8x4CGsO3+Hhxv8T/Q5PaXtkKtu6ug5TOeNV6gBiFeWPGFN9MuhOf23Q9Ifjh" crossorigin="anonymous">
        <link rel="stylesheet" href="https://cdn.jsdelivr.net/npm/bootstrap-icons@1.8.1/font/bootstrap-icons.css">
        <style>
            body { font-family: sans-serif; margin: 2em; background-color: #f8f9fa; }
            .card { margin-bottom: 1.5em; }
            .metric { font-size: 2.5em; font-weight: bold; color: #007bff; }
        </style>
    </head>
    <body>
        <div class="container">
            <div class="card text-center">
                <div class="card-header">
                    <h1><i class="bi bi-calendar-check" style="color: cornflowerblue;"></i> Your Vocabulary Report</h1>
                </div>
            </div>

            <div class="card">
                <div class="card-body">
                    <h2 class="card-title">Overall Progress</h2>
                    <p class="card-text">Total words added:</p>
                    <p class="metric">{{ total_words }}</p>
                </div>
            </div>

            <div class="card">
                 <div class="card-header">
                    Productivity
                </div>
                <ul class="list-group list-group-flush">
                    <li class="list-group-item">Most productive month: <strong>{{ most_productive_month }}</strong> ({{ max_words_in_month }} words)</li>
                    <li class="list-group-item">
                        <strong>Words per Month:</strong>
                        <ul class="list-unstyled mt-2">
                        {% for month, count in monthly_counts.items() %}
                            <li>{{ month }}: {{ count }} words</li>
                        {% endfor %}
                        </ul>
                    </li>
                </ul>
            </div>

            <div class="card">
                <div class="card-header">
                    Content Analysis
                </div>
                <div class="card-body">
                    <h5 class="card-title">Top 5 Parts of Speech:</h5>
                    <ul class="list-group">
                    {% for pos, count in pos_counts %}
                        <li class="list-group-item d-flex justify-content-between align-items-center">
                            {{ pos.capitalize() }}
                            <span class="badge badge-primary badge-pill">{{ count }}</span>
                        </li>
                    {% endfor %}
                    </ul>
                    <hr>
                    <h5 class="card-title mt-4">Words by Color Tag:</h5>
                    <ul class="list-group">
                    {% for color, count in color_counts.items() %}
                        <li class="list-group-item">
                            <i class="bi bi-square-fill" style="color: {{ color }}; margin-right: 8px;"></i>
                            {{ color.capitalize() }}: {{ count }} words
                        </li>
                    {% endfor %}
                    </ul>
                </div>
            </div>
        </div>
    </body>
    </html>
    """

    words = Word.objects()
    if not words:
        return "<p>No data available to generate a report. Start adding words!</p>"

    # Сбор статистики
    total_words = len(words)
    monthly_counts = Counter(word.date_added.strftime('%Y-%m') for word in words)
    most_productive_month = monthly_counts.most_common(1)[0][0] if monthly_counts else "N/A"
    max_words_in_month = monthly_counts.most_common(1)[0][1] if monthly_counts else 0
    pos_counts = Counter(word.part_of_speech for word in words).most_common(5)
    color_counts = Counter(word.color for word in words if word.color)

    from jinja2 import Template
    template = Template(JINJA_TEMPLATE)

    # Рендеринг шаблона Jinja
    return template.render(
        total_words=total_words,
        monthly_counts=dict(sorted(monthly_counts.items())),
        most_productive_month=most_productive_month,
        max_words_in_month=max_words_in_month,
        pos_counts=pos_counts,
        color_counts=dict(sorted(color_counts.items()))
    )
```

## Возможность использования Airflow

1. **Ежедневное "Слово дня":** Можно создать DAG в Airflow, который запускается каждый день.

    *  Задача 1: Выбирает случайное или самое старое неизученное слово из базы данных пользователя.
    * Задача 2: Отправляет это слово (с определением и примером) пользователю через E-mail или Telegram-бот. Это поможет регулярно повторять материал.
2. **Периодическая генерация отчётов:** Вместо того чтобы генерировать отчёт по запросу, Airflow может делать это автоматически (например, раз в месяц) и сохранять результат в PDF-файл в облачное хранилище или отправлять на email.

3. **Обогащение данных:** Можно настроить DAG, который периодически проходит по старым словам в базе и ищет для них дополнительную информацию (например, антонимы, этимологию, примеры из литературы), улучшая тем самым уже существующие записи.

4. **Резервное копирование:** Простой, но важный DAG, который раз в неделю делает бэкап базы данных AdvancedVocabDBAmina и сохраняет его в безопасное место.
